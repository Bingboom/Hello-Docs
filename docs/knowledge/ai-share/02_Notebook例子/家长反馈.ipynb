# 例子：从家长反馈中找出反复出现的问题

这是为本次分享编写的中文例子，全部使用虚构资料，不是对 GitHub 项目的翻译或改编。

先看各段代码下面的运行结果，再改“想看的问题”，看结果怎样变化。这里的反馈已经整理成固定栏目，主题也已人工确定；代码负责统计和查找，不会自动理解原始文章。

阅读：打开同文件夹里的“阅读版.html”，不用安装软件。
运行：用 Jupyter 打开本文件，从上到下运行。需要 Python 3 和 Jupyter；代码本身只使用 Python 自带功能。最后一步会在当前目录新建或更新 output/家长反馈摘要.md。

## 1. 先用这些资料

共有 6 条反馈，同一位学生可以有多条。先运行，看看程序读到了什么。

```python
反馈 = [
    {"学生": "张三", "主题": "作业安排", "内容": "希望提前知道本周作业安排。", "出处": "虚构反馈01"},
    {"学生": "李四", "主题": "阅读习惯", "内容": "希望提供适合课外阅读的书目。", "出处": "虚构反馈02"},
    {"学生": "王五", "主题": "作业安排", "内容": "想了解每天大致需要多少作业时间。", "出处": "虚构反馈03"},
    {"学生": "张三", "主题": "阅读习惯", "内容": "想了解怎样陪孩子讨论读过的书。", "出处": "虚构反馈04"},
    {"学生": "李四", "主题": "作业安排", "内容": "希望能集中查看本周作业要求。", "出处": "虚构反馈05"},
    {"学生": "王五", "主题": "家校沟通", "内容": "希望知道适合联系老师的时间。", "出处": "虚构反馈06"},
]
print(f"共读到 {len(反馈)} 条反馈")
for 记录 in 反馈:
    print(f"{记录['学生']}｜{记录['主题']}｜{记录['出处']}")
```

## 2. 看哪些问题出现得多

统计的是反馈条数，不是人数，也不自动代表问题的重要程度。哪些内容值得在家长会讲，由老师决定。

```python
from collections import Counter

统计 = Counter(记录["主题"] for 记录 in 反馈)
for 主题, 条数 in 统计.most_common():
    print(f"{主题}：{条数} 条反馈")
```

## 3. 改一个地方，再运行

先看“作业安排”的结果。把第一行中的“作业安排”改成“阅读习惯”，再运行这一段，就会看到另外两条反馈。试试“出勤”，看看没有记录时会怎样。

```python
想看的问题 = "作业安排"

选中的反馈 = [记录 for 记录 in 反馈 if 记录["主题"] == 想看的问题]
print(f"{想看的问题}：找到 {len(选中的反馈)} 条反馈")
if not 选中的反馈:
    print("暂无相关记录")
for 记录 in 选中的反馈:
    print(f"{记录['学生']}：{记录['内容']}（出处：{记录['出处']}）")
```

## 4. 把结果留下来

这一段会把刚才选中的内容写成一份文件。更改第三步后，再运行这一步，输出文件也会更新；原始反馈不变。

```python
from pathlib import Path

段落 = [f"# 家长反馈摘要：{想看的问题}", "", f"共 {len(选中的反馈)} 条反馈。", ""]
for 记录 in 选中的反馈:
    段落.append(f"- {记录['学生']}：{记录['内容']}（出处：{记录['出处']}）")
if not 选中的反馈:
    段落.append("暂无相关记录。")
文件 = Path("output") / "家长反馈摘要.md"
文件.parent.mkdir(exist_ok=True)
文件.write_text("\n".join(段落), encoding="utf-8")
print(f"已保存：{文件}")
print(文件.read_text(encoding="utf-8"))
```
